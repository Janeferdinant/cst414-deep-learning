Implementation of Rosenblatt Perceptron using pure NumPy and Matplotlib decision boundary plotting.


```python
import numpy as np
import matplotlib.pyplot as plt

class Perceptron:
    def __init__(self, input_dim, lr=0.1, epochs=20):
        self.w = np.zeros(input_dim)
        self.b = 0.0
        self.lr = lr
        self.epochs = epochs
        self.history = []

    def predict(self, x):
        z = np.dot(x, self.w) + self.b
        return 1 if z >= 0 else 0

    def fit(self, X, y):
        for epoch in range(self.epochs):
            errors = 0
            for xi, target in zip(X, y):
                pred = self.predict(xi)
                err = target - pred
                if err != 0:
                    self.w += self.lr * err * xi
                    self.b += self.lr * err
                    errors += 1
            self.history.append((self.w.copy(), self.b, errors))
            if errors == 0:
                print(f"Converged at epoch {epoch + 1}")
                break

# AND Gate Dataset
X = np.array([[0,0], [0,1], [1,0], [1,1]])
y = np.array([0, 0, 0, 1])

p = Perceptron(input_dim=2, lr=0.2)
p.fit(X, y)
print("Trained Weights:", p.w, "Bias:", p.b)
for xi, yi in zip(X, y):
    print(f"Input: {xi} -> Target: {yi}, Prediction: {p.predict(xi)}")
```
