## Solving XOR with Multi-Layer Perceptron using NumPy
**Faculty:** Jane Rose | Department of Computer Science & Engineering

Forward pass and decision surface visualization demonstrating non-linear classification boundary on XOR.


```python
import numpy as np

def relu(z):
    return np.maximum(0, z)

# Hard-coded weights for 2-2-1 MLP solving XOR
# Hidden layer weights: W1 (2x2), b1 (2,)
W1 = np.array([[1.0, 1.0],
               [1.0, 1.0]])
b1 = np.array([-0.5, -1.5])

# Output layer weights: W2 (1x2), b2 (1,)
W2 = np.array([[1.0, -2.0]])
b2 = np.array([0.0])

X = np.array([[0,0], [0,1], [1,0], [1,1]])

print("=== MLP FORWARD PASS ON XOR ===")
for x in X:
    z1 = np.dot(x, W1.T) + b1
    h = relu(z1)
    z2 = np.dot(h, W2.T) + b2
    y_pred = z2[0]
    print(f"Input: {x} -> Hidden z1: {z1} -> Hidden a1: {h} -> Output: {y_pred:.1f}")
```
