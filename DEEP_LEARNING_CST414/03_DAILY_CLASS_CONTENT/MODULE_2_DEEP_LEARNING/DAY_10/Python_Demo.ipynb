## Vanilla GD vs Momentum on an Anisotropic Quadratic Bowl
**Faculty:** Jane Rose | Department of Computer Science & Engineering

Trajectory comparison on an ill-conditioned quadratic function f(x, y) = 0.5 * (x^2 + 20*y^2).


```python
import numpy as np

# Quadratic bowl: steep in y, gentle in x
def cost(w):
    return 0.5 * (w[0]**2 + 20.0 * w[1]**2)

def grad(w):
    return np.array([w[0], 20.0 * w[1]])

w_init = np.array([-4.0, 2.0])
lr = 0.08

# 1. Vanilla Gradient Descent
w_gd = w_init.copy()
gd_traj = [w_gd.copy()]
for _ in range(15):
    g = grad(w_gd)
    w_gd -= lr * g
    gd_traj.append(w_gd.copy())

# 2. Momentum (beta = 0.9)
w_mom = w_init.copy()
v = np.zeros(2)
beta = 0.9
mom_traj = [w_mom.copy()]
for _ in range(15):
    g = grad(w_mom)
    v = beta * v - lr * g
    w_mom += v
    mom_traj.append(w_mom.copy())

print("After 15 steps:")
print(f"Vanilla GD Position: ({w_gd[0]:.4f}, {w_gd[1]:.4f}) | Loss: {cost(w_gd):.4f}")
print(f"Momentum   Position: ({w_mom[0]:.4f}, {w_mom[1]:.4f}) | Loss: {cost(w_mom):.4f}")
```
