## 2D Image Convolution from Scratch with NumPy
**Faculty:** Jane Rose | Department of Computer Science & Engineering

Pure NumPy implementation of 2D discrete cross-correlation applying Sobel edge detection kernels.


```python
import numpy as np

def conv2d(image, kernel):
    i_h, i_w = image.shape
    k_h, k_w = kernel.shape
    out_h = i_h - k_h + 1
    out_w = i_w - k_w + 1
    output = np.zeros((out_h, out_w))
    for i in range(out_h):
        for j in range(out_w):
            patch = image[i:i+k_h, j:j+k_w]
            output[i, j] = np.sum(patch * kernel)
    return output

image = np.array([
    [10, 10, 10, 0, 0],
    [10, 10, 10, 0, 0],
    [10, 10, 10, 0, 0],
    [10, 10, 10, 0, 0],
    [10, 10, 10, 0, 0]
])

sobel_v = np.array([[-1, 0, 1], [-2, 0, 2], [-1, 0, 1]])
feature_map = conv2d(image, sobel_v)
print("Convolved Feature Map (3x3):\n", feature_map)
```
